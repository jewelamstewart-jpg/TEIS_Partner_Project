We have some wonky tonky numbers in RDI columns, I bet there's more. Better look them up.

```python
for col in evals_df.columns:
    if evals_df[col].astype(str).str.contains(r'[A-Za-z]{3}-\d{2}', na=False).any():
        print(col)
```

```python
rdi = [col for col in evals_df.columns if 'rdi' in col.lower()]
```

```python
evals_df[rdi].head()
```

```python
evals_df[rdi].dtypes
```

```python
rdi_clean = [col for col in evals_df.columns if 'rdi' in col.lower()]

months = {
    'Jan-90': '1/90',
    'Feb-90': '2/90',
    'Mar-90': '3/90',
    'Apr-90': '4/90',
    'May-90': '5/90',
    'Jun-90': '6/90',
    'Jul-90': '7/90',
    'Aug-90': '8/90',
    'Sep-90': '9/90',
    'Oct-90': '10/90',
    'Nov-90': '11/90',
    'Dec-90': '12/90'}

for col in rdi_clean:
    evals_df[col] = evals_df[col].replace(months)
```
