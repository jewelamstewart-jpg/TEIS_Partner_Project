```python
evals_df.head()
```

First we'll have to calculate the age. So date of testing minus dob.

```python
evals_df['DOB'] = pd.to_datetime(evals_df['Date of Birth'])
evals_df['Assessment_Date'] = pd.to_datetime(evals_df['Adaptive-Self Care Date of Testing'])
```

```python
evals_df['age_years'] = (round((evals_df['Assessment_Date'] - evals_df['DOB']).dt.days / 365, 2))
evals_df['age_years'].head()
```

Some children are too young at testing for years, okay, we can do months. Oh, but months have different number of days...

```python
evals_df['age_months'] = (round((evals_df['Assessment_Date'] - evals_df['DOB']).dt.days / 30.55, 2))
```

```python
evals_df['age_months'].head()
```
